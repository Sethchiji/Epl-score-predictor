# 00 · One-time setup: put the starter repo on GitHub (no laptop storage needed)

Run this **once**. It takes the zip file you downloaded and pushes it to a new GitHub repo, entirely in the cloud.

### Before you run it (5 minutes)
1. **Create an empty repo on GitHub**: github.com → **New repository** → name it `epl-score-predictor` → **Public** → do **NOT** tick "Add a README" → **Create repository**.
2. **Create a token** so Colab can push to that repo: GitHub → profile picture → **Settings → Developer settings → Personal access tokens → Fine-grained tokens → Generate new token**.
   - Repository access: **Only select repositories** → `epl-score-predictor`
   - Permissions → Repository permissions → **Contents: Read and write**
   - Copy the token (starts with `github_pat_`). You only see it once.
3. **Save the token in Colab**: click the 🔑 **key icon** in the left sidebar → **Add new secret** → Name: `GITHUB_TOKEN`, Value: your token → turn on **Notebook access**.

Never paste the token directly into a code cell. Secrets keep it out of your notebook and out of GitHub.

```python
# ✏️ Fill these in
GITHUB_USER = "Sethchiji"            # your GitHub username
REPO_NAME   = "epl-score-predictor"
GIT_NAME    = "Chijioke Otuonye"      # shows on your commits
GIT_EMAIL   = "you@example.com"       # use the email on your GitHub account
```

```python
# Upload epl-score-predictor.zip when the button appears
from google.colab import files
uploaded = files.upload()
zip_name = next(iter(uploaded))
print("Uploaded:", zip_name)
```

```python
import os, subprocess
from google.colab import userdata

token = userdata.get("GITHUB_TOKEN")
remote = f"https://{token}@github.com/{GITHUB_USER}/{REPO_NAME}.git"

!rm -rf /content/{REPO_NAME} && unzip -q -o "{zip_name}" -d /content/
%cd /content/{REPO_NAME}
!git init -q -b main
!git config user.name "{GIT_NAME}"
!git config user.email "{GIT_EMAIL}"
!git add . && git commit -q -m "Starter repo: EPL score predictor (Phase 1)"
push = subprocess.run(["git", "push", "-q", remote, "main"], capture_output=True, text=True)
if push.returncode == 0:
    print(f"✅ Done! Open https://github.com/{GITHUB_USER}/{REPO_NAME}")
else:   # hide the token if git echoes the URL in its error
    print("❌ Push failed:", push.stderr.replace(token, "***"))
```
